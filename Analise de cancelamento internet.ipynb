# Análise de Dados com Python

### Desafio:

Você trabalha em uma empresa de telecom e tem clientes de vários serviços diferentes, entre os principais: internet e telefone.

O problema é que, analisando o histórico dos clientes dos últimos anos, você percebeu que a empresa está com Churn de mais de 26% dos clientes.

Isso representa uma perda de milhões para a empresa.

O que a empresa precisa fazer para resolver isso?

Base de Dados: https://drive.google.com/drive/folders/1T7D0BlWkNuy_MDpUHuBG44kT80EmRYIs?usp=sharing <br>
Link Original do Kaggle: https://www.kaggle.com/radmirzosimov/telecom-users-dataset

```python
#Importar e visualizar a base de dados
import pandas as pd
tabela = pd.read_csv("telecom_users.csv")
tabela = tabela.drop("Unnamed: 0", axis = 1)
display (tabela)
```

```python
#Tratamentos de dados
#tratar tipo de valor
tabela["TotalGasto"] = pd.to_numeric(tabela["TotalGasto"], errors ="coerce")
#deletar colunas vazias
tabela = tabela.dropna(how = "all" , axis = 1)
#deletar linhas vazias
tabela = tabela.dropna(how = "any" , axis = 0)
tabela.info()
```

```python
#Analise 
print(tabela["Churn"].value_counts())
print(tabela["Churn"].value_counts(normalize=True).map("{:.1%}".format))
```

```python
#comparar colunas da tabela com a coluna de cancelamento
import plotly.express as px
for coluna in tabela.columns: 
    grafico = px.histogram(tabela, x = coluna, color = "Churn")
    grafico.show()
```

### Conclusões e Ações


- Clientes que pagam mensalmente tem mais chance de cancelar,
      Fazer uma promoção mais agresiva para o plano anual.
- Clientes com pouco tempo de contrato tem mais chance de cancelar ,
    Primeira experiência com a operadora pode estar causando algum descontentamento com o cliente,
    Trazer algum incentivo para manter os clientes.
- Pessoas com filhos e casados tem menos chance de cancelar o plano,
    Focar em planos para familias.
- Oferecer mais serviços para os clientes,
    Planos mais acessiveis com mais serviços.
- Verificar a qualidade da fibra, a quantidade de cacelamento muito alo comparado ao modelo DSL.
- A forma de pagamento,
    Podemos oferecer algum desconto com o pagamento em debito automatico, dando a possibilidade dos clientes migrarem para essa opção.
